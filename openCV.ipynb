```python
import cv2
import numpy as np

img = cv2.imread("cat.jpg")

rows, cols = img.shape[:2]

# Four corners of the original image
src = np.float32([
    [0, 0],
    [cols-1, 0],
    [0, rows-1],
    [cols-1, rows-1]
])

# New positions of those corners
dst = np.float32([
    [50, 50],
    [cols-100, 0],
    [0, rows-50],
    [cols-50, rows-1]
])

# Compute the perspective transformation matrix
M = cv2.getPerspectiveTransform(src, dst)

# Apply the perspective transformation
warped = cv2.warpPerspective(img, M, (cols, rows))

cv2.imshow("Original", img)
cv2.imshow("Perspective", warped)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cat=cv2.imread("cat.jpg")
cat=cv2.cvtColor(cat,cv2.COLOR_BGR2RGB)
resized1=cv2.resize(cat,dsize=(800,500),fx=2,fy=0.5,interpolation=cv2.INTER_CUBIC)

plt.imshow(resized1)
plt.axis("off")
plt.title("INTER_CUBIC")
plt.show()
```
